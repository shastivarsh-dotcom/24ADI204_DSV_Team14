## 1. Load the Raw Dataset

```python
# Make sure AccidentsBig.csv is in the same folder as this notebook

df = pd.read_csv("AccidentsBig.csv", low_memory=False)

print("Dataset loaded successfully!")
```

## 2. Dataset Dimensions

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)
```

### Observation

The dataset contains a collection of accident records with multiple attributes describing accident, vehicle, road, environmental, and other related information.

## 3. Preview of the Dataset

```python
# First five records
df.head()
```

```python
# Last five records
df.tail()
```

## 4. Column Names

```python
for i, col in enumerate(df.columns, start=1):
    print(i, col)
```

## 5. Dataset Information and Data Types

```python
df.info()
```

### Observation

The `info()` output shows the number of records, column names, non-null counts, data types, and memory usage. The dataset contains a mixture of numerical and categorical/text attributes.

## 6. Numerical and Categorical Columns

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include=["object", "category"]).columns.tolist()

print("Numerical columns:")
for col in numerical_columns:
    print("-", col)

print("\nCategorical/Text columns:")
for col in categorical_columns:
    print("-", col)
```

## 7. Missing-Value Representation

```python
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
}).sort_values("Missing Values", ascending=False)

missing_summary
```

### Observation

Missing values are only being **identified and represented** in Week 2. No imputation or deletion is performed at this stage. Detailed missing-value handling will be part of the data-cleaning stage.

## 8. Duplicate Record Check

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

## 9. Descriptive Statistics

```python
df.describe().T
```

### Observation

Descriptive statistics provide the count, mean, standard deviation, minimum, quartiles, and maximum values of numerical attributes. These values give an initial understanding of the scale and distribution of the dataset.

## 10. Unique Values in Important Columns

```python
# Display the number of unique values for each column

unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique(dropna=True) for col in df.columns]
})

unique_summary
```

## 11. Accident Severity Representation

```python
if "Accident_Severity" in df.columns:
    severity_counts = df["Accident_Severity"].value_counts(dropna=False)
    print(severity_counts)
else:
    print("Column 'Accident_Severity' was not found in this dataset.")
```

```python
if "Accident_Severity" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x="Accident_Severity")
    plt.title("Accident Severity Distribution")
    plt.xlabel("Accident Severity")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=15)
    plt.show()
```

### Observation

The count plot provides a basic representation of the frequency of the accident-severity categories. Detailed interpretation and relationships with other variables will be performed during later EDA.
